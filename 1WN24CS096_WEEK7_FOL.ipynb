```python
def is_variable(x):
    return isinstance(x, str) and x[0].islower()

def substitute(x, subs):
    while isinstance(x, str) and x in subs:
        x = subs[x]
    if isinstance(x, tuple):
        return tuple(substitute(i, subs) for i in x)
    return x

def occurs(var, term, subs):
    term = substitute(term, subs)
    if var == term:
        return True
    if isinstance(term, tuple):
        return any(occurs(var, i, subs) for i in term)
    return False

def unify(a, b, subs):
    a = substitute(a, subs)
    b = substitute(b, subs)

    if a == b:
        return True

    if is_variable(a):
        if occurs(a, b, subs):
            return False
        subs[a] = b
        return True

    if is_variable(b):
        if occurs(b, a, subs):
            return False
        subs[b] = a
        return True

    if isinstance(a, tuple) and isinstance(b, tuple):
        if len(a) != len(b) or a[0] != b[0]:
            return False
        for x, y in zip(a[1:], b[1:]):
            if not unify(x, y, subs):
                return False
        return True

    return False

def parse(s):
    s = s.strip()
    if '(' not in s:
        return s

    name, rest = s.split('(', 1)
    rest = rest[:-1]
    args = []
    depth = 0
    start = 0

    for i, ch in enumerate(rest):
        if ch == '(':
            depth += 1
        elif ch == ')':
            depth -= 1
        elif ch == ',' and depth == 0:
            args.append(parse(rest[start:i]))
            start = i + 1

    args.append(parse(rest[start:]))
    return tuple([name.strip()] + args)

e1 = input("Enter first expression: ")
e2 = input("Enter second expression: ")

subs = {}

if unify(parse(e1), parse(e2), subs):
    print("Unification Successful")
    print("Substitutions:", subs)
else:
    print("Unification Failed")

print("\n 1WN24CS096 DISHA BAID")
```
