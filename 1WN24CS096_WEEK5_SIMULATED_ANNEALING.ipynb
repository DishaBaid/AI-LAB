**SIMULATED ANNEALING TO SOLVE 8-QUEENS**

```python
import random
import math

def heuristic(state):
    conflicts = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts


def simulated_annealing():
    current = [random.randint(0, 7) for _ in range(8)]
    temperature = 100
    cooling_rate = 0.99

    while temperature > 0.01:
        if heuristic(current) == 0:
            return current

        neighbor = current.copy()
        col = random.randint(0, 7)
        neighbor[col] = random.randint(0, 7)

        delta = heuristic(current) - heuristic(neighbor)

        if delta > 0 or random.random() < math.exp(delta / temperature):
            current = neighbor

        temperature *= cooling_rate

    return current if heuristic(current) == 0 else None


def print_board(state):
    if state is None:
        print("No solution found!")
        return

    for row in range(8):
        print(" ".join("Q" if state[col] == row else "."
                       for col in range(8)))


solution = simulated_annealing()

if solution:
    print("Solution:", solution)
    print("Conflicts:", heuristic(solution))
    print("\nChessboard:")
    print_board(solution)
else:
    print("Solution not found. Run again.")

print("\nDISHA BAID 1WN24CS096")
```
